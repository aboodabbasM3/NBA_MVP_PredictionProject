```python
#Creating a list of the years I wanna work on
years = list(range(1991,2026))
```

```python
#The website I wanna webscrape
url_start = "https://www.basketball-reference.com/awards/awards_{}.html"
```

### Initially I am going to web scrape and get data for all the players who were MVP contenders

```python
#Web scraping and downloading all html files for each year to my folder so I dont keep redownloading everytime
#Initially i just tried using regular pull requests but I kept getting blocked from bot detection, so i switched to selenium
from selenium import webdriver
import time
import os

os.makedirs("MVPs", exist_ok=True)

driver = webdriver.Chrome()

for year in years:
    url = url_start.format(year)
    driver.get(url)
    
    time.sleep(5)  # let Cloudflare + page load
    
    html = driver.page_source
    
    with open(f"MVPs/{year}.html", "w", encoding="utf-8") as f:
        f.write(html)
    
    print(f"Saved {year}")

driver.quit()
```

```python
#Using beautiful soup to parse my HTML files that I saved
from bs4 import BeautifulSoup
```

```python
#Importing pandas because we want to turn the tables into dataframes for future machine learning
import pandas as pd

#Importing StringIO to convert an HTML string into a file for pandas
from io import StringIO
```

```python
#Creating an empty list for each years dataframe
dfs = []

#Looping through each year to pull our html files from the folder to parse the html and turn it into a dataframe
for year in years:
    with open("MVPs/{}.html".format(year)) as f:
        page = f.read()

    #Intializing a beautiful soup class with the page string we made earlier as an html parser 
    soup = BeautifulSoup(page, "html.parser")

    #Removing the repeated header rows in the tables
    soup.find('tr',class_="over_header").decompose()

    #Finding the MVP tables by using its HTML ID
    mvptable = soup.find(id="mvp")

    #Turning the table into a dataframe
    mvp = pd.read_html(StringIO(str(mvptable)))[0]

    #Adding a year column to be able to identify the years later when we combine all the dataframes
    mvp["Year"] = year

    dfs.append(mvp)
```

```python
#Combining the list of dataframes into a single dataframe
mvps = pd.concat(dfs)
```

```python
#Turning the dataframe into a csv for future use
mvps.to_csv("mvps.csv")
```

### Now I am going to webscrape and get the stats for all the players for each season

```python
playerstats_url = "https://www.basketball-reference.com/leagues/NBA_{}_per_game.html"

#Repeating the same steps I used to webscrape earlier
os.makedirs("PlayerStats", exist_ok=True)

driver = webdriver.Chrome()

for year in years:
    url = playerstats_url.format(year)
    driver.get(url)
    
    time.sleep(5)  # let Cloudflare + page load
    
    html = driver.page_source
    
    with open(f"PlayerStats/{year}.html", "w", encoding="utf-8") as f:
        f.write(html)
    
    print(f"Saved {year}")

driver.quit()
```

```python
dfs = []

#Again I am going to loop through each year to pull our html files from the folder to parse the html and turn it into a dataframe
for year in years:
    with open("PlayerStats/{}.html".format(year)) as f:
        page = f.read()

    soup = BeautifulSoup(page, "html.parser")
    soup.find('tr',class_="thead").decompose()
    playertable = soup.find(id="per_game_stats")
    playerstats = pd.read_html(StringIO(str(playertable)))[0]
    playerstats["Year"] = year

    dfs.append(playerstats)
```

```python
players = pd.concat(dfs)
```

```python
#Turning our data into a csv for future ML
players.to_csv("PlayerStats.csv")
```

### Lastly I am going to web scrape all team data, this will be useful in helping me predict the MVP

```python
#Scraping the data
team_stats_url = "https://www.basketball-reference.com/leagues/NBA_{}_standings.html"

for year in years:
    url = team_stats_url.format(year)
    driver.get(url)
    
    time.sleep(5)  # let Cloudflare + page load
    
    html = driver.page_source
    
    with open(f"Teams/{year}.html", "w", encoding="utf-8") as f:
        f.write(html)
        
    print(f"Saved {year}")
    
```

```python
#Parsing the data with beautiful soup
dfs = []

#Again I am going to loop through each year to pull our html files from the folder to parse the html and turn it into a dataframe
for year in years:
    with open("Teams/{}.html".format(year)) as f:
        page = f.read()

    soup = BeautifulSoup(page, "html.parser")
    soup.find('tr',class_="thead").decompose()
    teamstable = soup.find(id="divs_standings_E")
    teamstable = pd.read_html(StringIO(str(teamstable)))[0]
    teamstable["Year"] = year
    teamstable["Team"] = teamstable["Eastern Conference"]
    del teamstable["Eastern Conference"]

    dfs.append(teamstable)

    soup = BeautifulSoup(page, "html.parser")
    soup.find('tr',class_="thead").decompose()
    teamstable = soup.find(id="divs_standings_W")
    teamstable = pd.read_html(StringIO(str(teamstable)), flavor="lxml")[0]
    teamstable["Year"] = year
    teamstable["Team"] = teamstable["Western Conference"]
    del teamstable["Western Conference"]

    dfs.append(teamstable)
```

```python
teams = pd.concat(dfs)
```

### Cleaning the MVP dataset

```python
import pandas as pd
```

```python
mvps = pd.read_csv("mvps.csv")
```

```python
#Keeping only the neccessary columns
mvps = mvps[["Player", "Year", "Pts Won", "Pts Max", "Share"]]
```

```python
mvps
```

### Cleaning the player stats dataset

```python
players = pd.read_csv("PlayerStats.csv")
```

```python
players
```

```python
#Removing unneeded columns
del players["Unnamed: 0"]
del players["Rk"]
del players["Awards"]
```

```python
#Some players have multiple rows for the same season since they played for different teams
#To handle this i will group by player and year and keep only one row per season which will be the average of their season stats
#Denoted in the dataset as 3TM

players.groupby(["Player", "Year"])

def singlerow(df):
    if df.shape[0]==1:
        return df
    else:
        row = df[df["Team"] == "3TM"]
        row["Team"] = df.iloc[-1,:]["Team"]
        return row

players = players.groupby(["Player", "Year"]).apply(singlerow)
```

```python
players.head(50)
```

```python
players.index = players.index.droplevel()
```

```python
players.index = players.index.droplevel()
```

```python
players
```

#### Now we can merge the two datasets

```python
#Here i am combining the MVP and players dataframes on the player and year columns
combined = players.merge(mvps, how= "outer", on= ["Player", "Year"])
```

```python
combined.head()
```

```python
#There are many players in our dataframe that didnt recieve any MVP votes so they have NAN, for data cleaning purposes
#I will fill those NANs with zeros because they realistically got no votes

combined[["Pts Won", "Pts Max", "Share"]] = combined[["Pts Won", "Pts Max", "Share"]].fillna(0)
```

```python
combined.head()
```

### Now I will clean the Teams dataset

```python
teams = pd.read_csv("Teams.csv")
```

```python
teams.head(30)
```

```python
#From our webscraping there are some rows that are invalid and are just called Midwest Division or Pacific Division

#This basically get rids of any rows where the win column has the word division
teams = teams[~teams["W"].str.contains("Division")]
```

```python
#Some teams also have asterisks by their names if they made the playoffs so we have to make that consistent

teams["Team"] = teams["Team"].str.replace("*", "", regex= False)
```
